## Standardise Column Names

Get a list of column names.

We will standardise the column names into snake_case using the .rename() function.

```python
# code taken from Gemini
rename_columns = {
    'car_ID': 'car_id',
    'CarName': 'car_name',
    'fueltype': 'fuel_type',
    'doornumber': 'door_number',
    'carbody': 'body_style',
    'drivewheel': 'drive_wheel',
    'enginelocation': 'engine_location',
    'wheelbase': 'wheel_base',
    'carlength': 'car_length',
    'carwidth': 'car_width',
    'carheight': 'car_height',
    'curbweight': 'curb_weight',
    'enginetype': 'engine_type',
    'cylindernumber': 'cylinder_number',
    'enginesize': 'engine_size',
    'fuelsystem': 'fuel_system',
    'boreratio': 'bore_ratio',
    'compressionratio': 'compression_ratio',
    'peakrpm': 'peak_rpm',
    'citympg': 'city_mpg',
    'highwaympg': 'highway_mpg'
}

df = df.rename(columns=rename_columns)
```

```python
df.columns
```

## Transform Column Values

We want to standardise the names inside `car_name`, separating the brand from the model and search for typos. We will first do a value counts to see if all entries have the brand name first.

```python
df.value_counts('car_name')
```

The brand name comes first. We can loop throguh the entries in this column, grab the item at index 0 and create a new column, `brand`.

```python
# Gemini used for code structure and the split function.
df['brand'] = [car.split()[0] for car in df['car_name']]
    
```

```python
df['brand'].value_counts()
```

Calling value counts above allows us to see misspellings. We will now replace the misspellings, then call value counts again to check everything is correct.

```python
# Gemini code used
brand_replacements = {
    'alfa-romero': 'alfa-romeo',
    'vw': 'volkswagen',
    'vokswagen': 'volkswagen',
    'maxda': 'mazda',
    'toyouta': 'toyota',
    'Nissan': 'nissan',
    'porcshce': 'porsche'
}

df['brand'] = df['brand'].replace(brand_replacements)
```

```python
df['brand'].value_counts()
```

We will repeat these steps for the columns `cylinder_number` and `door_number` to convert their string numbers to integers.

```python
df['door_number'].value_counts()
```

```python
# changed Gemini's code
door_replacements = {
    'four': 4,
    'two': 2
}

df['door_number'] = df['door_number'].replace(door_replacements)
```

```python
# check that the data type has changed to int64
df['door_number'].info()
```

```python
df['cylinder_number'].value_counts()
```

```python
# changed Gemini's code and used copilot predictions to complete the code below
cylinder_replacements = {
    'four': 4,
    'six': 6,
    'five': 5,
    'eight': 8,
    'two': 2,
    'three': 3,
    'twelve': 12,
}

df['cylinder_number'] = df['cylinder_number'].replace(cylinder_replacements)
```

```python
df['cylinder_number'].info()
```

The data type for both of these columns has been converted to int64. We see that there are a total of 205 entries, so every row in the dataframe has a value in these columns.

## Duplicate Check

AI Note: Generative AI was used to assist in phrasing refinement and structure formatting of the below text.

We will do a final check for duplicates to make sure that there were no hidden duplicates earlier due to the messy strings with typos in the column `car_name`. 

- Checking for duplicates including `car_name` would fail to catch identical cars entered with different spelling variations.
- Dropping `car_name` may result in false positives, as two distinct models could theoretically share identical specifications.

To address this, we will use a two-step validation process:
1. **Specification Screen:** Check for duplicate rows across all technical specifications while omitting `car_name`.
2. **Conditional Inspection:** 
   - If 0 matches are found, it confirms no two listings share identical specs, ruling out duplicate records entirely.
   - If matches are found, filter and manually inspect those candidate rows to verify whether they represent the same vehicle or different models.

```python
df.drop(columns=['car_name']).duplicated().sum()
```

There are no duplicates with the same specs, so we do not need to continue to step 2.
